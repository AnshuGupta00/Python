## Vectot Matrix and Tensor

```python
arr=np.array([1,2,3,4])
print("The value of arr are :",arr)# Array is Vector


matrix=np.array([[1,2,3,4],[5,6,7,9]])
print("The value of Matrix are :",matrix)# This is Tensor


tensor=np.array([[[1,2,3,5],[5,6,9,4],
                  [2,3,6,9],[9,6,8,5]]])# Tensor has multidiminsional
print("The value of Tensor array are:",tensor)
```

## Arryay Properties

```python
arr=np.array([[23,56,96,85],
             [85,96,95,82]])
print("Shape",arr.shape)
print("Size",arr.size)
print("Dimensio",arr.ndim)
print("Dtype",arr.dtype)
```
